Resume Summarization

```python
# ==========================================
# RESUME SUMMARY USING SUMY
# ==========================================

from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lsa import LsaSummarizer
import nltk
nltk.download('punkt_tab')

# ==========================================
# SAMPLE RESUME TEXT
# ==========================================

resume_text = """
Experienced Python developer skilled in Flask,
Machine Learning, SQL, REST APIs, and NLP.

Worked on AI-based recruitment systems,
resume analyzers, and web applications.

Strong understanding of backend development,
semantic similarity, and data processing.

Completed multiple AI and machine learning projects.
"""

# ==========================================
# CREATE PARSER
# ==========================================

parser = PlaintextParser.from_string(
    resume_text,
    Tokenizer("english")
)

# ==========================================
# LOAD SUMMARIZER
# ==========================================

summarizer = LsaSummarizer()

# ==========================================
# GENERATE SUMMARY
# ==========================================

summary = summarizer(
    parser.document,
    2   # number of summary sentences
)

# ==========================================
# PRINT SUMMARY
# ==========================================

print("\n========== RESUME SUMMARY ==========\n")

for sentence in summary:
    print(sentence)
```

ATS Score  using TF-IDF 

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Resume text
resume = """
Python developer skilled in Flask,
Machine Learning, SQL, REST APIs
"""

# Job description
job_description = """
Looking for a Python developer with Flask,
SQL and Machine Learning experience
"""

# TF-IDF vectorization
vectorizer = TfidfVectorizer()

vectors = vectorizer.fit_transform([resume, job_description])

# Similarity score
similarity = cosine_similarity(vectors[0:1], vectors[1:2])

# Convert into ATS %
ats_score = round(similarity[0][0] * 100, 2)

print("ATS Score:", ats_score)
```

ATS score(sentence-tansformers, done)

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load pretrained semantic model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Resume text
resume = """
Python developer skilled in Flask,
Machine Learning, SQL, REST APIs
"""

# Job description
job_description = """
Looking for a Python developer with Flask,
SQL and Machine Learning experience
"""

# Convert text into embeddings
resume_embedding = model.encode([resume])
jd_embedding = model.encode([job_description])

# Calculate semantic similarity
similarity = cosine_similarity(
    resume_embedding,
    jd_embedding
)

# ATS score
ats_score = round(similarity[0][0] * 100, 2)

print("ATS Score:", ats_score)
```
